```python
import os
import geopandas as gpd
```

```python
# Define the folder containing the KFS data files
# Since the notebook is located in the "notebooks" directory/folder,
# we need to go up one level(..) to access the "data/raw/kfs" folder.

folder = "../data/raw/kfs"

# Verify that Python can locate the data folder and the files within it

# Shows where the notebook is running from
print("Current directory:", os.getcwd())  
# Checks whether the folder exists   
print("Folder exists:", os.path.exists(folder))
# Lists all files inside the folder
print("Files:", os.listdir(folder))

# List of KFS datasets to inspect

files = [
    "kfs_raw.geojson",
    "protected_area.geojson",
    "kenya_gazetted_forest_-2241218727256807108.geojson",
    "Kenya_Forest_Type_2010_Dataset.geojson"
]

# Loop through each dataset and inspect its contents

for file in files:

    # Create the full file path
    path = os.path.join(folder, file)

    # Display a separator for readability
    print("\n" + "=" * 80)

    # Print the name of the dataset being inspected
    print("Reading:", path)
    
    # Check whether the file exists before reading it
    print("Exists?", os.path.exists(path))

    # Load the GeoJSON file into a GeoDataFrame
    gdf = gpd.read_file(path)

    # Display the number of records(rows)
    print("Rows:", len(gdf))

    # Display all available columns
    print("Columns:")
    print(gdf.columns.tolist())
```

#### Deep-Diviving into the kfs_raw.geojson

##### Step 1: Load the dataset

```python
import geopandas as gpd

# Load the dataset
kfs_raw = gpd.read_file("../data/raw/kfs/kfs_raw.geojson")
```

##### Step 2: General information

```python
# Dataset dimensions
print("Shape:", kfs_raw.shape)

# Column names
print("\nColumns:")
print(kfs_raw.columns.tolist())

# Data types
print("\nData Types:")
print(kfs_raw.dtypes)
```

##### Step 3: Preview the data

```python
# Display the first 10 rows
kfs_raw.head(10)
```

##### Step 4: Check for missing values

```python
print("\nMissing Values:")
print(kfs_raw.isnull().sum())
```

##### Step 5: Check duplicate forest names

```python
print("Unique forest names:", kfs_raw["FORESTS_"].nunique())

print("\nDuplicate forest names:")

duplicates = kfs_raw["FORESTS_"].value_counts()

print(duplicates[duplicates > 1].head(20))
```

##### Step 6: Inspect the **LU** column

```python
# Most important part

print("Unique Land Use values:")

print(kfs_raw["LU"].value_counts(dropna=False))
```

##### Step 7: Area statistics

```python
print(kfs_raw["AREA"].describe())
```

##### Step 8: Geometry types

```python
print(kfs_raw.geometry.geom_type.value_counts())
```

##### Step 9: CRS(Coordiate Reference System)

```python
print(kfs_raw.crs)
```

##### Step 10: Sample forest names

```python
print(kfs_raw["FORESTS_"].sort_values().head(30))
```

##### Findings of kfs_raw.geojson

- Every polygon is a **multipolygon**
- CRS is already **EPSG:4326**
- There are **no forest names**

So this dataset is basically a collection of forest boundaries, **not a list of named forests**.

Therefore, the cleaned dataset only needs:
- geometry
- area_sq_km
- source
