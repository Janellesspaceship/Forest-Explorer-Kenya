# Forest Explorer Kenya
## Exploratory Data Analysis (EDA) of GBIF Wildlife Data

This notebook explores wildlife occurrence data downloaded from the GBIF API.
The goal is to understand the dataset before cleaning and transforming it for the Forest Explorer Kenya application.

## 1. Import Required Libraries

```python
import json
import pandas as pd
```

## 2. Load the GBIF Dataset

```python

with open("../data/raw/gbif/gbif_test.json", "r", encoding="utf-8") as file:
    data = json.load(file)

df = pd.DataFrame(data["results"])
```

## 5. Selecting Relevant Columns

The GBIF dataset contains many fields. For the Forest Explorer application,
we only keep the columns needed for mapping wildlife observations.

```python
locations = df[[
    "scientificName",
    "country",
    "countryCode",
    "stateProvince",
    "verbatimLocality",
    "decimalLatitude",
    "decimalLongitude"
]]

locations.head()
```

## 6. Missing Values in Selected Columns

```python
locations.isnull().sum()
```
